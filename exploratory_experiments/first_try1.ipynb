```python
from openai import OpenAI

client = OpenAI(
    api_key="Input your key",
    base_url="Input your URL"
)

def response_API(user_prompt: str) -> str:
    """
    Call GPT API and return model response text
    """
    try:
        response = client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[
                {"role": "user", "content": user_prompt}
            ],
            temperature=0
        )


        return response.choices[0].message.content

    except Exception as e:
        print("API call failed:", e)
        return None
```

```python
# Test gpt4o


user_prompt = f"""
Claim: Says Nancy Pelosi was arrested after ripping a copy of the State of the Union speech.

Question:
Can you provide a detailed explanation?
"""
result = response_API(user_prompt)
print(result)
```
